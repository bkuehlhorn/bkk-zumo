```python
from pyvis.network import Network

net = Network(
    notebook=True,
    directed=True
)
```

```python
# net.add_node("Singapore")
# net.add_node("San Francisco")
# net.add_node("Tokyo")
# net.add_nodes(["Riga", "Copenhagen"],
#               color=['lightgreen', 'yellow'])

states = ["init", "state1", "state2", "state3", "state4","done"]
          # "state5", "state6", "state7",
net.add_nodes(states)
```

```python

stateMatrix = {
    "init": {
        "init": "init",
        "event1": "state1",
        "event2": "state2",
        "event3": "state3",
        "event4": "state4",
        "timeout": "init",
        "done": "done"
    },
    "state1": {
        "init": "state1",
        "event1": "init",
        "event2": "state2",
        "event3": "state3",
        "event4": "state4",
        "timeout": "init",
        "done": "done"
    },
    "state2": {
        "init": "state2",
        "event1": "state1",
        "event2": "init",
        "event3": "state3",
        "event4": "state4",
        "timeout": "init",
        "done": "done"
    },
    "state3": {
        "init": "state3",
        "event1": "state1",
        "event2": "state2",
        "event3": "init",
        "event4": "state4",
        "timeout": "init",
        "done": "done"
    },
    "state4": {
        "init": "state4",
        "event1": "state1",
        "event2": "state2",
        "event3": "init",
        "event4": "state4",
        "timeout": "init",
        "done": "done"
    },
    "done": {
        "init": "state4",
        "event1": "state1",
        "event2": "state2",
        "event3": "init",
        "event4": "state4",
        "timeout": "init",
        "done": "done"
    },
}
```

```python
for sm_keys in stateMatrix.keys():
    for sm_event in stateMatrix[sm_keys].keys():
        net.add_edge(sm_keys, stateMatrix[sm_keys][sm_event], title=sm_event)

# net.show('mygraph.html')
```

```python
net.repulsion(
    spring_length=400,
    damping=0.01,
)
net.show('mygraph.html')
```
